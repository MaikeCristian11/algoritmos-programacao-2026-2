---## 2. A anatomia do for```pythonfor numero in range(1, 6):    print(numero)```- `numero` é a **variável de controle**. Ela é criada pelo próprio `for` e recebe um valor diferente em cada repetição.- `range(1, 6)` é o **conjunto de valores** que ela vai assumir: 1, 2, 3, 4 e 5.- os **dois pontos** e o **recuo** funcionam igual ao `if`: o recuo marca o que está dentro do laço.Cada volta do laço tem um nome: **iteração**. O exemplo acima tem cinco iterações.

### O range não inclui o último númeroEsta é a maior fonte de erro do conteúdo de hoje. `range(1, 6)` vai até 5, e não até 6.Rode a célula abaixo e compare as três saídas.

```````python
print("range(5):")for i in range(5):    print(i)print()print("range(1, 5):")for i in range(1, 5):    print(i)print()print("range(1, 10, 2):")for i in range(1, 10, 2):    print(i)
```````

Três formas de escrever:| Escrita | Significa | Valores ||---|---|---|| `range(5)` | de 0 até 4 | 0, 1, 2, 3, 4 || `range(1, 5)` | de 1 até 4 | 1, 2, 3, 4 || `range(1, 10, 2)` | de 1 até 9, pulando de 2 em 2 | 1, 3, 5, 7, 9 |Quando você quiser contar de 1 a N, escreva `range(1, N + 1)`.**Mude e rode de novo:** faça a célula acima mostrar os números pares de 2 a 10.

---## 4. AcumuladorO `soma` do exemplo anterior tem nome: é um **acumulador**. Ele guarda o resultado parcial e vai crescendo a cada iteração.Duas regras que valem sempre:- o acumulador começa em **zero**- ele é inicializado **antes** do laço, nunca dentroRode a célula abaixo, que mostra o que acontece quando essa segunda regra é quebrada.

```````python
print("certo:")soma = 0for i in range(1, 6):    soma = soma + iprint(soma)print()print("errado:")for i in range(1, 6):    soma = 0    soma = soma + iprint(soma)
```````

No caso errado, o acumulador volta para zero em cada iteração, então no fim ele guarda apenas o último valor.Esse erro não gera mensagem nenhuma. Só aparece no teste de mesa ou quando o resultado não faz sentido.

---## 5. ContadorO contador é parecido com o acumulador, mas em vez de somar valores, ele conta quantas vezes algo aconteceu. Ele também começa em zero e cresce de um em um.Aqui o `for` e o `if` trabalham juntos: o laço passa por todas as notas e o `if` decide se aquela nota entra na contagem.

```````python
aprovados = 0for i in range(1, 6):    nota = float(input(f"Nota do aluno {i}: "))    if nota >= 7:        aprovados = aprovados + 1print("Aprovados:", aprovados)
```````
