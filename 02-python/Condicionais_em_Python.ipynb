---## 2. A estrutura do if```pythonif condição:    o que fazer quando for verdadeiraelse:    o que fazer quando for falsa```Três detalhes que o Python cobra:- os **dois pontos** no fim da linha do `if` e do `else`- o **recuo** de quatro espaços nas linhas de dentro. É ele que marca onde o bloco começa e termina- a condição precisa ser uma pergunta de sim ou nãoO `else` é opcional. Sem ele, quando a condição é falsa o algoritmo simplesmente não faz nada e segue adiante.

````python
idade = 20if idade >= 18:    print("Pode dirigir")print("Fim do programa")
````

Os operadores de comparação| Pergunta | Símbolo | Exemplo ||---|---|---|| é maior que | `>` | `nota > 7` || é menor que | `<` | `idade < 18` || é maior ou igual | `>=` | `media >= 7` || é menor ou igual | `<=` | `faltas <= 15` || é igual a | `==` | `resposta == "sim"` || é diferente de | `!=` | `senha != "1234"` |**Atenção ao igual.** Um sinal só (`=`) guarda um valor na variável. Dois sinais (`==`) comparam. É o erro que mais vai aparecer nas próximas semanas.No VisuAlg é o contrário do que vocês conhecem: lá a comparação usa um sinal só.

````python
# guardar: um sinalx = 10# comparar: dois sinaisprint(x == 10)print(x == 7)print(x != 7)
````

````python
nota = float(input("Digite a sua nota: "))if nota >= 7:    print("Aprovado")else:    print("Reprovado")
````

---## 5. Mais de dois caminhosQuando a resposta tem três ou mais possibilidades, entra o `elif`, que é a abreviação de "senão, se".O Python testa as condições de cima para baixo e para na primeira que for verdadeira. As de baixo nem chegam a ser avaliadas.

````python
media = float(input("Media do aluno: "))if media >= 7:    print("Aprovado")elif media >= 5:    print("Recuperacao")else:    print("Reprovado")
````

**A ordem importa.** Se as duas primeiras condições fossem trocadas de lugar, um aluno com média 8 cairia no `media >= 5` e apareceria em recuperação.Teste: troque as duas primeiras linhas de lugar e rode com 8.
